# Logs Evaluation

Notebook για evaluation του `log_compliance_checker` πάνω σε 100 synthetic procurement logs και 12 targeted regulations.

```python
import json
import os

from logs_evaluation_utils import run_logs_evaluation

REGULATIONS_JSON_PATH = "extracted_regulations_CELEX.json"
OUTPUT_ROOT = "Evaluation Results/logs_evaluiation"
MODEL = "gpt-5-mini"
API_KEY = ""

if not API_KEY:
    raise RuntimeError("Set OPENAI_API_KEY before running this notebook.")
```

```python
result = run_logs_evaluation(
    regulations_json_path=REGULATIONS_JSON_PATH,
    model=MODEL,
    api_key=API_KEY,
    output_root=OUTPUT_ROOT,
    positive_variants_per_rule=5,
    compliant_case_count=40,
    verbose=True,
)

manifest = result["manifest"]
summary = result["summary_metrics"]

print(json.dumps(manifest, ensure_ascii=False, indent=2))
print(json.dumps({
    "overall_pair_metrics": summary["overall_pair_metrics"],
    "overall_case_metrics": summary["overall_case_metrics"],
}, ensure_ascii=False, indent=2))
```
